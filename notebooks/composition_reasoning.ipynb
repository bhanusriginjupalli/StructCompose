```python
import json
import networkx as nx
import numpy as np
import pandas as pd
```

```python
graph_path = r"C:\Users\bhanu\Downloads\StructCompose\data\graphs\sample_scg.json"

with open(graph_path, "r") as f:

    graph_data = json.load(f)

G = nx.node_link_graph(graph_data)

print(G)
```

```python
for node, data in G.nodes(data=True):

    print(node)
    print(data)
    print()
```

```python
third_scores = []

for node, data in G.nodes(data=True):

    d = data["thirds_distance"]

    score = 1 / (d + 1)

    third_scores.append(score)

overall_thirds_score = np.mean(third_scores)

print("Rule of Thirds Score:", overall_thirds_score)
```

```python
balance_values = []

for u, v, data in G.edges(data=True):

    balance_values.append(data["balance"])

overall_balance = np.mean(balance_values)

print("Overall Balance:", overall_balance)
```

```python
tension_values = []

for u, v, data in G.edges(data=True):

    tension_values.append(data["tension"])

overall_tension = np.mean(tension_values)

print("Overall Tension:", overall_tension)
```

```python
weights = []

for node, data in G.nodes(data=True):

    weights.append(data["visual_weight"])

max_idx = np.argmax(weights)

dominant_node = list(G.nodes())[max_idx]

print("Dominant Subject:")

print(G.nodes[dominant_node])
```

```python
x_positions = []

for node, data in G.nodes(data=True):

    x_positions.append(data["center_x"])

spread = np.std(x_positions)

print("Spatial Spread:", spread)
```

```python
composition_profile = {

    "thirds_score": float(overall_thirds_score),

    "balance_score": float(overall_balance),

    "tension_score": float(overall_tension),

    "spatial_spread": float(spread)

}

composition_profile
```

```python
feedback = []
```

```python
if overall_thirds_score > 0.02:

    feedback.append(
        "Subjects align well with rule-of-thirds intersections."
    )

else:

    feedback.append(
        "Subjects are weakly aligned with thirds composition."
    )
```

```python
if overall_balance > 0.9:

    feedback.append(
        "Composition appears visually balanced and stable."
    )

else:

    feedback.append(
        "Frame exhibits asymmetric visual balance."
    )
```

```python
if overall_tension > 0.001:

    feedback.append(
        "Strong compositional tension creates dynamic framing."
    )

else:

    feedback.append(
        "Composition feels calm with low visual tension."
    )
```

```python
if spread > 250:

    feedback.append(
        "Subjects are spatially dispersed, emphasizing environmental scale."
    )

else:

    feedback.append(
        "Subjects are compositionally clustered."
    )
```

```python
print("COMPOSITION ANALYSIS\n")

for f in feedback:

    print("-", f)
```

```python
profile_df = pd.DataFrame([composition_profile])

profile_df.to_csv(

    r"C:\Users\bhanu\Downloads\StructCompose\data\metadata\composition_reasoning.csv",

    index=False
)

print("Saved reasoning profile")
```
